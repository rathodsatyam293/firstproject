```python
def sum_digits(num):
    total = 0
    while num > 0:
        total += num % 10
        num //= 10
    return total

def reverse_number(num):
    rev = 0
    while num > 0:
        rev = rev * 10 + (num % 10)
        num //= 10
    return rev

# Using 12345
num = 12345
print("Sum of digits:", sum_digits(num))
print("Reversed number:", reverse_number(num))
```

```python
# 4.
def find_larger(num1, num2):
  if num1 > num2:
    return num1
  else:
    return num2

# Demonstrate the function
number1 = 20
number2 = 35

larger_number = find_larger(number1, number2)
print(f"The larger number between {number1} and {number2} is: {larger_number}")
```

```python
# 6.
def count_vowel(string):
  counter = 0
  vowel = "AEIOUaeiou"
  for ch in string:
    if ch in vowel:
      counter+=1
  return counter

print(count_vowel("my name is satyam"))
```

```python
# 7.
def find_largest(numbers):
  largest = numbers[0]

  for num in numbers:
    if num>largest:
      largest = num
  return largest

print(find_largest([10,25,7,99,42]))
```
